# DS Lab 8: Data Visualization using Seaborn
Titanic Dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style('whitegrid')
```

```python
# Load Titanic dataset
dataset = sns.load_dataset('titanic')
dataset.head()
```

```python
dataset.info()
dataset.describe()
```

## Histogram of Fare (Required)

```python
sns.histplot(dataset['fare'], bins=10, kde=False)
plt.title("Fare Distribution")
plt.show()
# Insight: Most passengers paid low fares, while only a few paid very high ticket prices.
```

## Distribution Plot

```python
sns.histplot(dataset['age'], bins=10, kde=True)
plt.title("Age Distribution")
plt.show()
# Insight: Most passengers were young adults, with fewer very young children and elderly passengers.
```

## Joint Plot

```python
sns.jointplot(x='age', y='fare', data=dataset, kind='scatter')
plt.show()
# Insight: Higher fares appear across adult ages, while many low fares are spread across all ages.
```

## Rug Plot

```python
sns.rugplot(dataset['fare'])
plt.title("Rug Plot of Fare")
plt.show()
# Insight: Fare values are densely packed at lower prices, showing few passengers at expensive fare levels.
```

## Bar Plot

```python
sns.barplot(x='sex', y='age', data=dataset)
plt.title("Average Age by Gender")
plt.show()
# Insight: Male passengers have a slightly higher average age than female passengers.
```

## Count Plot

```python
sns.countplot(x='sex', data=dataset)
plt.title("Count of Gender")
plt.show()
# Insight: Male passengers are more numerous than female passengers in this dataset.
```

## Box Plot

```python
sns.boxplot(x='sex', y='age', data=dataset)
plt.title("Age Distribution by Gender")
plt.show()
# Insight: Male ages show a wider spread and more outliers than female ages.
```

## Violin Plot

```python
sns.violinplot(x='sex', y='age', data=dataset)
plt.title("Violin Plot")
plt.show()
# Insight: Both genders center around young adult ages, with males showing broader age distribution.
```

## Strip Plot

```python
sns.stripplot(x='sex', y='age', data=dataset, jitter=True)
plt.title("Strip Plot")
plt.show()
# Insight: Most points cluster below age 40, with overlap between male and female age values.
```

## Swarm Plot

```python
sns.swarmplot(x='sex', y='age', data=dataset)
plt.title("Swarm Plot")
plt.show()
# Insight: Age values overlap by gender, but males appear slightly more spread across older ages.
```

## Heatmap

```python
corr = dataset.corr(numeric_only=True)
sns.heatmap(corr, annot=True)
plt.title("Correlation Heatmap")
plt.show()
# Insight: Fare relates moderately with class-linked variables, while age shows generally weak correlations.
```
