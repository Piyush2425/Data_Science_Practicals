## Box Plot on Titanic Dataset

### Aim:
Plot a box plot for distribution of age with respect to gender and survival status in the Titanic dataset.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_style('whitegrid')
```

```python
dataset = sns.load_dataset('titanic')
dataset.head()
```

```python
dataset.info()
```

## Box Plot: Age vs Gender

```python
sns.boxplot(x='sex', y='age', data=dataset)
plt.title("Age Distribution by Gender")
plt.show()
```

**Insight:** Male passengers show wider age variation compared to female passengers.

## Box Plot: Age vs Gender and Survival

```python
sns.boxplot(x='sex', y='age', data=dataset, hue='survived')
plt.title("Age Distribution by Gender and Survival")
plt.show()
```

**Insight:** Younger passengers, especially females, had higher survival rates compared to older passengers.

## Conclusion
The box plot visualizes age distribution across gender and survival categories, highlighting median differences and outliers.
