## Iris Dataset Analysis

### Aim:
Load the Iris dataset, analyze its features, and perform visualizations using histograms and box plots to identify distributions and outliers.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris

sns.set_style('whitegrid')
```

```python
# Load Iris dataset
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)

# Add species column
df['species'] = iris.target

df.head()
```

```python
# Data information
df.info()
df.describe()
```

## Histograms of Features

```python
df.hist(figsize=(10,8))
plt.suptitle("Feature Distributions (Histograms)")
plt.show()
```

**Insight:** Petal features show clearer separation between species compared to sepal features.

## Box Plot (All Features)

```python
sns.boxplot(data=df.drop('species', axis=1))
plt.title("Box Plot of Iris Features")
plt.show()
```

**Insight:** Some outliers are visible, especially in sepal width, while petal features are more consistent.

## Improved Box Plot (Species vs Petal Length)

```python
sns.boxplot(x='species', y='petal length (cm)', data=df)
plt.title("Petal Length Distribution Across Species")
plt.show()
```

**Insight:** Petal length clearly separates species, making it a strong feature for classification.

## Conclusion
The Iris dataset visualization shows that petal-related features provide better separation between species and help in identifying patterns and outliers.
